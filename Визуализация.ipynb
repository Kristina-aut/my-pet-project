```python
import pandas as pd
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
sns.set_style('darkgrid')
```

```python
df = pd.read_csv('/content/insurance.csv')
```

```python
df
```

```python
iris = sns.load_dataset('iris')
sns.pairplot(iris)
plt.show()
```

```python
sns.pairplot(
data=iris, # датафрейм pandas
hue='species', # переменная для цветового кодирования
vars=['sepal_length', '' 'sepal_width', 'petal_length'], # выбор переменных
diag_kind='kde', # тип диагонального графика
height=2.5, # размер каждого подграфика
aspect=1.2, # соотношение сторон
plot_kws={'alpha': 0.6}, # параметры для scatter plots
diag_kws={'alpha': 0.8} # параметры для диагональных графиков
)
plt.suptitle('Анализ отношений', y=1.02)
plt.show()
```

```python
sns.pairplot(
data=df, # датафрейм pandas
hue='smoker', # переменная для цветового кодирования
vars=['age', 'bmi', 'charges'], # выбор переменных
diag_kind='kde', # тип диагонального графика
height=2.5, # размер каждого подграфика
aspect=1.2, # соотношение сторон
plot_kws={'alpha': 0.6}, # параметры для scatter plots
diag_kws={'alpha': 0.8} # параметры для диагональных графиков
)
plt.suptitle('Анализ отношений', y=1.02)
plt.show()
```

```python
sns.pairplot(
    data = df,
    hue = 'smoker',
    vars = ['age', 'bmi', 'children', 'charges'],
    diag_kind = 'kde',
    height = 2.5,
    aspect = 1.2,
    plot_kws = {'alpha': 0.6},
    diag_kws = {'alpha': 0.8}
)
plt.suptitle('Анализ зависимостей', y=1.02)
plt.show()
```

```python
sns.set(style = 'whitegrid', context = 'notebook')
plot = sns.pairplot(
    data = df,
    hue = "smoker",
    diag_kind='kde',
    palette="Set1",
    plot_kws={'alpha': 0.6},
    height=2.5
)
plot._legend.set_title("Курильщик?")
plt.show()
```

```python
plt.figure(figsize = (10, 6))
```

```python
sns.scatterplot(
    data=df,
    x='age',
    y='charges',
    hue='smoker',
    style='smoker',
    palette={'yes': 'red', 'no': 'blue'},
    s=80,
    alpha=0.7
)
plt.title('Зависимость расходов от возраста с учетом курения', fontsize = 14)
plt.xlabel('Возраст', fontsize=12)
plt.ylabel('Медицинские расходы', fontsize=12)
plt.legend(title='Курильщик?', loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

```python
smoker_data=df[df['smoker'] == 'yes']
```

```python
sns.lmplot(
    data=smoker_data,
    x='age',
    y='charges',
    hue='sex',
    palette={'female': 'magenta', 'male': 'green'},
    ci=95,
    scatter_kws={'alpha': 0.6, 's': 50},
    height=6,
    aspect=1.5
)
plt.title('Зависимость расходов от возраста для курящих', fontsize = 14)
plt.xlabel('Возраст', fontsize=12)
plt.ylabel('Медицинские расходы', fontsize=12)
plt.legend(title='Курильщик?', loc='upper left')
plt.grid(True, linestyle='--', alpha=0.3)

plt.show()

plt.savefig('Пилот.png')
```

```python
plt.figure(figsize = (10, 6))
sns.scatterplot(
    data=df,
    x='bmi',
    y='charges',
    hue='sex',
    style='sex',
    palette={'female': 'magenta', 'male': 'green'},
    s=70,
    alpha=0.6
)
plt.title('Зависимость расходов от индекса массы тела с учетом пола', fontsize = 14)
plt.xlabel('ИМТ (кг/м2)')
plt.ylabel('Медицинские расходы')
plt.legend(title='Пол')
plt.grid(True, linestyle='--', alpha=0.3)

plt.show()
```
